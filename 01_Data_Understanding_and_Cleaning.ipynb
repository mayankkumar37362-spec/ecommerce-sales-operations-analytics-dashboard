The missing values in the timestamp columns are not random.

After analyzing the `order_status` column, it was observed that these NULL values mainly belong to orders with statuses such as:

• canceled
• unavailable
• created
• processing/approved (orders not yet delivered)

Since these orders never reached later stages of the order lifecycle, the missing timestamps are expected and represent valid business information rather than data quality issues.

# Phase 3: Data Quality Report

```python
# this function is used to quality check the data

def data_quality_report(df):
    report = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.values,
        "Missing Values": df.isnull().sum().values,
        "Missing %": ((df.isnull().sum()/len(df))*100).round(2).values,
        "Unique Values": df.nunique().values
    })

    return report
```

## Missing Value Assessment

The missing values were classified into three categories:

1. Business Missing
   - Occur because of the business process.
   - Example: Delivery timestamps for cancelled orders.
   - Decision: Retained.

2. Optional Missing
   - Fields that users are not required to fill.
   - Example: Review comments.
   - Decision: Retained.

3. Data Quality Missing
   - Missing due to incomplete or incorrect data.
   - Example: Product dimensions.
   - Decision: To be investigated and handled appropriately.

Why did you verify the relationships before merging?"

"Before joining the tables, I validated the cardinality using the data rather than relying solely on documentation. This helped identify one-to-many relationships that could duplicate records during joins. By understanding the relationships first, I was able to choose the correct merge strategy and avoid incorrect business metrics."

```python
# orders if our fact table so want to keep every record of it so we are using left join and orders_df as our left table

analytics_df = orders_df.merge(
    customers_df,
    on="customer_id",
    how="left"
)
```

```python
print("Rows:", analytics_df.shape[0])

print("Columns:", analytics_df.shape[1])

print("Duplicates:", analytics_df.duplicated().sum())

print("Missing Values:")
print(analytics_df.isnull().sum())
```

```python
analytics_df = analytics_df.merge(
    order_items_df,
    on = "order_id",
    how = "left",
    validate="one_to_many"
    )
```

```python
payments_summary = (
    payments_df
    .groupby("order_id")
    .agg(
        total_payment=("payment_value", "sum"),
        payment_count=("payment_sequential", "count"),
        primary_payment_type=("payment_type", "first"),
        max_installments=("payment_installments", "max")
    )
    .reset_index()
)
```

```python
payments_summary.head()
```

```python
payments_summary.rename(columns={
    "total_payment": "payment_amount",
    "payment_count": "no_of_payments",
    "primary_payment_type": "payment_method",
    "max_installments": "installments"
}, inplace=True)
```

```python
analytics_df = analytics_df.merge(
    payments_summary,
    on="order_id",
    how="left",
    validate="many_to_one"
)
```

```python
print(analytics_df.shape)
analytics_df[[
    "payment_amount",
    "payment_method",
    "installments"
]].isnull().sum()
```

```python
reviews_summary = (
    reviews_df
    .groupby("order_id")
    .agg(
        review_score=("review_score", "mean"),
        review_count=("review_id", "count"),
        review_creation=("review_creation_date", "min")
    )
    .reset_index()
)
```

```python
analytics_df = analytics_df.merge(
    reviews_summary,
    on="order_id",
    how="left",
    validate="many_to_one"
)
```

```python
print(analytics_df.shape)
analytics_df[[
    "review_score",
    "review_count"
]].isnull().sum()
```

```python
analytics_df = analytics_df.merge(
    category_translation_df,
    on="product_category_name",
    how="left",
    validate="many_to_one"
)
```

```python
analytics_df.rename(
    columns={
        "product_category_name_english":"product_category"
    },
    inplace=True
)
```
